---

##### A ) Distância Euclidiana

$$d(p, q) = \sqrt{\sum_{i=1}^{n} (q_i - p_i)^2}$$

- | Matriz P e Q |
-> | Diferença entre Q e P |
-> | Quadrado da diferença |
-> | Somar todos os termos |
-> | Tirar a raiz |

```python
d_euclidean_value = distance.euclidean([10], [100])
```

```python
def d_euclidean(P : ArrayLike, Q : ArrayLike) -> float:
    return float(np.sqrt(np.sum(np.square(np.asarray(Q) - np.asarray(P)))))

d_euclidean(P, Q)
```

---

##### B ) Distância Chebyshev

$$D(p,q) = \max_{i} \left( |p_{i}-q_{i}| \right)$$

- | Matriz P e Q | -> | Diferença Entre Q e P | -> | Valor Máximo |

```python
value_d_chebyshev = distance.chebyshev(u=[10], v=[100])
```

```python
def d_chebyshev(P : ArrayLike, Q : ArrayLike) -> float:
    return float(np.max(np.abs(np.asarray(P) - np.asarray(Q)))) 

d_chebyshev(P, Q)
```

---

##### C ) Distância Manhattan

$$d_{1}(p, q) = \sum_{i=1}^{n} |p_i - q_i|$$

- | Matrizes P e Q | -> | Diferença Entre P e Q | -> | Somatório i até n |

```python
value_d_manhattan = distance.minkowski([10], [100], p=3)
```

```python
def d_manhattan(P : ArrayLike, Q : ArrayLike) -> float:
    return float(np.sum(np.abs(np.asarray(P) - np.asarray(Q)))) 

d_manhattan(P, Q)
```

```python
class knnClassifier:

    def __init__(self, n_neighbors : int = 3, distance_metric : str ='euclidean'):
        self.n_neighbors = n_neighbors,
        self.distance_metric = distance_metric.lower()

    def fit(self, X: ArrayLike, y: ArrayLike):
        X = np.asarray(X)
        y = np.asarray(y)
        
        if X.shape[0] != y.shape[0]:
            raise ValueError("X e y devem ter o mesmo número de amostras.")
        
        self.dimensional = np.column_stack((X, y))

    def predict(self, z):
        l = []
        z = np.asarray(z)

        if self.dimensional[:, 0:2][0].shape != z.shape:
            raise ValueError("x ∈ X e y devem ter o mesmo número de dimensões.")
        
        for x, y in zip(self.dimensional[:, 0:2], self.dimensional[:, 2]):
            l.append([d_euclidean(P=x, Q=z), int(y)])
        print(np.sort(l)[:3])
```

```python
X = np.random.randint(0, 100, size=(100, 2))
y = np.random.randint(0, 2, size=100)

print(f"| Shape X: {X.shape} \n| Shape y: {y.shape}")
```

```python
knn = knnClassifier(n_neighbors=3)
knn.fit(X, y)
knn.dimensional[:, 2]
knn.predict([1,1])
```
